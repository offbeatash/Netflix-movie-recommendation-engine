```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = pd.read_parquet("/workspaces/Netflix-movie-recommendation-engine/data/movie_genre_features.parquet")

print("Shape:", movies.shape)
print("\nColumns:")
print(movies.columns.tolist())
```

```python
print("\nFirst 5 movies:")
print(movies.head())
```

```python
#identify genre feature columns
genre_columns = movies.columns[2:]

print("\nNumber of genre features:", len(genre_columns))
print("\nGenre features:")
print(genre_columns.tolist())
```

```python
#separate movie information from genre features
movie_ids = movies["Movie_ID"]
movie_titles = movies["Title"]
```

```python
X = movies[genre_columns]

#calculate similarity between every pair of movies
similarity_matrix = cosine_similarity(X)

print("\nSimilarity matrix shape:", similarity_matrix.shape)
print("\nSimilarity example:")
print(similarity_matrix[0][:10])
```

```python
movie_name = "Lord of the Rings: The Fellowship of the Ring"

movie_index = movies.index[
    movies["Title"] == movie_name
][0]

print("Movie index:", movie_index)
print("Movie ID:", movie_ids.iloc[movie_index])
print("Movie:", movie_titles.iloc[movie_index])
```

```python
similarities = similarity_matrix[movie_index]

print(similarities[:20])
```
